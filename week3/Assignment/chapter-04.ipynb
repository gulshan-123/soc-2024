# Baseline strategies

```python
# This function implements a pure exploitation strategy over a specified number of episodes in a given environment.
# It initializes Q-values and counts for actions to zero, then iteratively selects actions with the highest Q-value,
# updates Q-values based on received rewards, and tracks the evolution of Q-values, returns, and actions taken across episodes.

def pure_exploitation(env, n_episodes=1000):
    # Initialize Q-values and counts for each action to zero
    Q = np.zeros((env.action_space.n), dtype=np.float64) 
    N = np.zeros((env.action_space.n), dtype=np.int)

    # Arrays to store Q-values, returns, and actions for each episode
    Qe = np.empty((n_episodes, env.action_space.n), dtype=np.float64)
    returns = np.empty(n_episodes, dtype=np.float64)
    actions = np.empty(n_episodes, dtype=np.int)
    name = 'Pure exploitation'

    # Loop through each episode
    for e in tqdm(range(n_episodes), 
                  desc='Episodes for: ' + name, 
                  leave=False):
        # Select the action with the highest Q-value
        action = np.argmax(Q)

        # Execute the action and observe the reward
        _, reward, _, _ = env.step(action)
        # Update the count for the selected action
        N[action] += 1
        # Update the Q-value for the selected action using the incremental formula
        Q[action] = Q[action] + (reward - Q[action])/N[action]
        
        # Store the current Q-values, reward, and action for this episode
        Qe[e] = Q
        returns[e] = reward
        actions[e] = action
    # Return the strategy name, per-episode returns, Q-values, and actions
    return name, returns, Qe, actions
```

```python
# This function implements a pure exploration strategy over a specified number of episodes in a given environment.
# Unlike pure exploitation, it selects actions randomly, allowing for exploration of the action space without bias towards the current highest Q-value.
# It updates Q-values based on received rewards, similar to the exploitation strategy, but ensures a diverse experience by exploring different actions.

def pure_exploration(env, n_episodes=1000):
    # Initialize Q-values and counts for each action to zero
    Q = np.zeros((env.action_space.n), dtype=np.float64)
    N = np.zeros((env.action_space.n), dtype=np.int)

    # Arrays to store Q-values, returns, and actions for each episode
    Qe = np.empty((n_episodes, env.action_space.n), dtype=np.float64)
    returns = np.empty(n_episodes, dtype=np.float64)
    actions = np.empty(n_episodes, dtype=np.int)
    name = 'Pure exploration'

    # Loop through each episode
    for e in tqdm(range(n_episodes), 
                  desc='Episodes for: ' + name, 
                  leave=False):
        # Select an action randomly
        action = np.random.randint(len(Q))
        
        # Execute the action and observe the reward
        _, reward, _, _ = env.step(action)
        # Update the count for the selected action
        N[action] += 1
        # Update the Q-value for the selected action using the incremental formula
        Q[action] = Q[action] + (reward - Q[action])/N[action]
        
        # Store the current Q-values, reward, and action for this episode
        Qe[e] = Q
        returns[e] = reward
        actions[e] = action
    # Return the strategy name, per-episode returns, Q-values, and actions
    return name, returns, Qe, actions
```

# Simple strategies

```python
# This function implements the epsilon-greedy strategy for a given environment over a specified number of episodes.
# It balances exploration and exploitation by selecting the best action (exploitation) with probability 1-epsilon
# and a random action (exploration) with probability epsilon. This approach allows for both learning from the best-known actions
# and discovering potentially better actions by exploring.

def epsilon_greedy(env, epsilon=0.01, n_episodes=1000):
    # Initialize Q-values and counts for each action to zero
    Q = np.zeros((env.action_space.n), dtype=np.float64)
    N = np.zeros((env.action_space.n), dtype=np.int)

    # Arrays to store Q-values, returns, and actions for each episode
    Qe = np.empty((n_episodes, env.action_space.n), dtype=np.float64)
    returns = np.empty(n_episodes, dtype=np.float64)
    actions = np.empty(n_episodes, dtype=np.int)
    name = 'Epsilon-Greedy {}'.format(epsilon)

    # Loop through each episode
    for e in tqdm(range(n_episodes), 
                  desc='Episodes for: ' + name, 
                  leave=False):
        # Decide between exploration and exploitation
        if np.random.uniform() > epsilon:
            # Exploitation: select the action with the highest Q-value
            action = np.argmax(Q)
        else:
            # Exploration: select a random action
            action = np.random.randint(len(Q))

        # Execute the action and observe the reward
        _, reward, _, _ = env.step(action)
        # Update the count for the selected action
        N[action] += 1
        # Update the Q-value for the selected action using the incremental formula
        Q[action] = Q[action] + (reward - Q[action])/N[action]
        
        # Store the current Q-values, reward, and action for this episode
        Qe[e] = Q
        returns[e] = reward
        actions[e] = action
    # Return the strategy name, per-episode returns, Q-values, and actions
    return name, returns, Qe, actions
```

```python
def lin_dec_epsilon_greedy(env,
                           init_epsilon=1.0,
                           min_epsilon=0.01, 
                           decay_ratio=0.05, 
                           n_episodes=1000):
    Q = np.zeros((env.action_space.n), dtype=np.float64)
    N = np.zeros((env.action_space.n), dtype=np.int)

    Qe = np.empty((n_episodes, env.action_space.n), dtype=np.float64)
    returns = np.empty(n_episodes, dtype=np.float64)
    actions = np.empty(n_episodes, dtype=np.int)
    name = 'Lin Epsilon-Greedy {}, {}, {}'.format(init_epsilon, 
                                          min_epsilon, 
                                          decay_ratio)
    for e in tqdm(range(n_episodes), 
                  desc='Episodes for: ' + name, 
                  leave=False):
        decay_episodes = n_episodes * decay_ratio
        epsilon = 1 - e / decay_episodes
        epsilon *= init_epsilon - min_epsilon
        epsilon += min_epsilon
        epsilon = np.clip(epsilon, min_epsilon, init_epsilon)
        if np.random.uniform() > epsilon:
            action = np.argmax(Q)
        else:
            action = np.random.randint(len(Q))

        _, reward, _, _ = env.step(action)
        N[action] += 1
        Q[action] = Q[action] + (reward - Q[action])/N[action]
        
        Qe[e] = Q
        returns[e] = reward
        actions[e] = action
    return name, returns, Qe, actions
```

```python
# This function implements an exponentially decaying epsilon-greedy strategy for a given environment over a specified number of episodes.
# It starts with a high epsilon for more exploration and exponentially decays the epsilon value to encourage more exploitation as learning progresses.
# This approach allows the agent to explore widely in the beginning and then gradually focus more on exploiting the best-known actions as it learns.

def exp_dec_epsilon_greedy(env, 
                           init_epsilon=1.0,
                           min_epsilon=0.01,
                           decay_ratio=0.1,
                           n_episodes=1000):
    # Initialize Q-values and counts for each action to zero
    Q = np.zeros((env.action_space.n), dtype=np.float64)
    N = np.zeros((env.action_space.n), dtype=np.int)

    # Arrays to store Q-values, returns, and actions for each episode
    Qe = np.empty((n_episodes, env.action_space.n), dtype=np.float64)
    returns = np.empty(n_episodes, dtype=np.float64)
    actions = np.empty(n_episodes, dtype=np.int)

    # Calculate the epsilon values for each episode, decaying exponentially
    decay_episodes = int(n_episodes * decay_ratio)
    rem_episodes = n_episodes - decay_episodes
    epsilons = np.logspace(-2, 0, decay_episodes) * (init_epsilon - min_epsilon)
    epsilons += min_epsilon
    epsilons = np.pad(epsilons, (0, rem_episodes), 'edge')
    name = 'Exp Epsilon-Greedy {}, {}, {}'.format(init_epsilon, 
                                                  min_epsilon, 
                                                  decay_ratio)
    # Loop through each episode
    for e in tqdm(range(n_episodes), 
                  desc='Episodes for: ' + name, 
                  leave=False):
        # Decide between exploration and exploitation based on the current epsilon
        if np.random.uniform() > epsilons[e]:
            # Exploitation: select the action with the highest Q-value
            action = np.argmax(Q)
        else:
            # Exploration: select a random action
            action = np.random.randint(len(Q))

        # Execute the action and observe the reward
        _, reward, _, _ = env.step(action)
        # Update the count for the selected action
        N[action] += 1
        # Update the Q-value for the selected action using the incremental formula
        Q[action] = Q[action] + (reward - Q[action])/N[action]
        
        # Store the current Q-values, reward, and action for this episode
        Qe[e] = Q
        returns[e] = reward
        actions[e] = action
    # Return the strategy name, per-episode returns, Q-values, and actions
    return name, returns, Qe, actions
```

```python
# This function implements the optimistic initialization strategy for a given environment over a specified number of episodes.
# It initializes Q-values and counts with optimistic estimates to encourage exploration of all actions.
# Actions are selected based on the highest Q-value, which initially favors unexplored or less explored actions due to their optimistic initial values.

def optimistic_initialization(env, 
                              optimistic_estimate=1.0,
                              initial_count=100,
                              n_episodes=1000):
    # Initialize Q-values and counts with optimistic estimates
    Q = np.full((env.action_space.n), optimistic_estimate, dtype=np.float64)
    N = np.full((env.action_space.n), initial_count, dtype=np.int)
    
    # Arrays to store Q-values, returns, and actions for each episode
    Qe = np.empty((n_episodes, env.action_space.n), dtype=np.float64)
    returns = np.empty(n_episodes, dtype=np.float64)
    actions = np.empty(n_episodes, dtype=np.int)
    name = 'Optimistic {}, {}'.format(optimistic_estimate, 
                                      initial_count)
    # Loop through each episode
    for e in tqdm(range(n_episodes), 
                  desc='Episodes for: ' + name, 
                  leave=False):
        # Select the action with the highest Q-value
        action = np.argmax(Q)

        # Execute the action and observe the reward
        _, reward, _, _ = env.step(action)
        # Update the count for the selected action
        N[action] += 1
        # Update the Q-value for the selected action using the incremental formula
        Q[action] = Q[action] + (reward - Q[action])/N[action]

        # Store the current Q-values, reward, and action for this episode
        Qe[e] = Q
        returns[e] = reward
        actions[e] = action
    # Return the strategy name, per-episode returns, Q-values, and actions
    return name, returns, Qe, actions
```

# Running simple strategies on Two-Armed Bandit environments

```python
def b2_run_simple_strategies_experiment(env_name='BanditTwoArmedUniform-v0'):
    results = {}
    experiments = [
        # baseline strategies
        lambda env: pure_exploitation(env),
        lambda env: pure_exploration(env),

        # epsilon greedy
        lambda env: epsilon_greedy(env, epsilon=0.07),
        lambda env: epsilon_greedy(env, epsilon=0.1),

        # epsilon greedy linearly decaying
        lambda env: lin_dec_epsilon_greedy(env, 
                                           init_epsilon=1.0,
                                           min_epsilon=0.0,
                                           decay_ratio=0.1),
        lambda env: lin_dec_epsilon_greedy(env, 
                                           init_epsilon=0.3,
                                           min_epsilon=0.001,
                                           decay_ratio=0.1),

        # epsilon greedy exponentially decaying
        lambda env: exp_dec_epsilon_greedy(env, 
                                           init_epsilon=1.0, 
                                           min_epsilon=0.0,
                                           decay_ratio=0.1),
        lambda env: exp_dec_epsilon_greedy(env, 
                                           init_epsilon=0.3, 
                                           min_epsilon=0.0, 
                                           decay_ratio=0.3),

        # optimistic
        lambda env: optimistic_initialization(env, 
                                              optimistic_estimate=1.0, 
                                              initial_count=10),
        lambda env: optimistic_initialization(env, 
                                              optimistic_estimate=1.0, 
                                              initial_count=50),
    ]
    for env_seed in tqdm(SEEDS, desc='All experiments'):
        env = gym.make(env_name, seed=env_seed) ; env.reset()
        true_Q = np.array(env.env.p_dist * env.env.r_dist)
        opt_V = np.max(true_Q)
        for seed in tqdm(SEEDS, desc='All environments', leave=False):
            for experiment in tqdm(experiments, 
                                   desc='Experiments with seed {}'.format(seed), 
                                   leave=False):
                env.seed(seed) ; np.random.seed(seed) ; random.seed(seed)
                name, Re, Qe, Ae = experiment(env)
                Ae = np.expand_dims(Ae, -1)
                episode_mean_rew = np.cumsum(Re) / (np.arange(len(Re)) + 1)
                Q_selected = np.take_along_axis(
                    np.tile(true_Q, Ae.shape), Ae, axis=1).squeeze()
                regret = opt_V - Q_selected
                cum_regret = np.cumsum(regret)
                if name not in results.keys(): results[name] = {}
                if 'Re' not in results[name].keys(): results[name]['Re'] = []
                if 'Qe' not in results[name].keys(): results[name]['Qe'] = []
                if 'Ae' not in results[name].keys(): results[name]['Ae'] = []
                if 'cum_regret' not in results[name].keys(): 
                    results[name]['cum_regret'] = []
                if 'episode_mean_rew' not in results[name].keys(): 
                    results[name]['episode_mean_rew'] = []

                results[name]['Re'].append(Re)
                results[name]['Qe'].append(Qe)
                results[name]['Ae'].append(Ae)
                results[name]['cum_regret'].append(cum_regret)
                results[name]['episode_mean_rew'].append(episode_mean_rew)
    return results

b2_results_s = b2_run_simple_strategies_experiment()
```

# Advanced strategies

```python
# This function implements the softmax action selection strategy for a given environment over a specified number of episodes.
# It uses a temperature parameter that decays over time to balance exploration and exploitation.
# Initially, actions are chosen almost randomly, but as the temperature decreases, the strategy increasingly favors actions with higher Q-values.

def softmax(env, 
            init_temp=float('inf'), 
            min_temp=0.0,
            decay_ratio=0.04,
            n_episodes=1000):
    Q = np.zeros((env.action_space.n), dtype=np.float64)
    N = np.zeros((env.action_space.n), dtype=np.int)

    Qe = np.empty((n_episodes, env.action_space.n), dtype=np.float64)
    returns = np.empty(n_episodes, dtype=np.float64)
    actions = np.empty(n_episodes, dtype=np.int)
    name = 'Lin SoftMax {}, {}, {}'.format(init_temp, 
                                           min_temp,
                                           decay_ratio)
    # Adjust initial temperature if it's set to infinity
    init_temp = min(init_temp, sys.float_info.max)
    # Ensure minimum temperature is not zero to avoid division by zero
    min_temp = max(min_temp, np.nextafter(np.float32(0), np.float32(1)))

    for e in tqdm(range(n_episodes), desc='Episodes for: ' + name, leave=False):
        # Calculate the current temperature using linear decay
        temp = max(min_temp, init_temp - (init_temp - min_temp) * (e / (n_episodes * decay_ratio)))
        
        # Calculate the softmax probabilities
        scaled_Q = Q / temp
        exp_Q = np.exp(scaled_Q - np.max(scaled_Q))  # Subtract max for numerical stability
        probs = exp_Q / np.sum(exp_Q)
        
        # Select action based on the calculated probabilities
        action = np.random.choice(np.arange(len(probs)), p=probs)

        # Execute the action and observe the reward
        _, reward, _, _ = env.step(action)
        # Update the count and Q-value for the selected action
        N[action] += 1
        Q[action] = Q[action] + (reward - Q[action]) / N[action]
        
        # Store the current Q-values, reward, and action for this episode
        Qe[e] = Q
        returns[e] = reward
        actions[e] = action

    return name, returns, Qe, actions
```

```python
# This function implements the Upper Confidence Bound (UCB) action selection strategy for a given environment over a specified number of episodes.
# It balances exploration and exploitation by considering not only the estimated rewards (Q-values) but also the uncertainty or variance in those estimates.
# The parameter c controls the degree of exploration, with higher values favoring exploration over exploitation.

def upper_confidence_bound(env, 
                           c=2, 
                           n_episodes=1000):
    # Initialize Q-values and counts for each action to zero
    Q = np.zeros((env.action_space.n), dtype=np.float64)
    N = np.zeros((env.action_space.n), dtype=np.int)
    
    # Arrays to store Q-values, returns, and actions for each episode
    Qe = np.empty((n_episodes, env.action_space.n), dtype=np.float64)
    returns = np.empty(n_episodes, dtype=np.float64)
    actions = np.empty(n_episodes, dtype=np.int)
    name = 'UCB {}'.format(c)
    
    for e in tqdm(range(n_episodes), desc='Episodes for: ' + name, leave=False):
        # For the first few episodes, select actions sequentially to ensure all actions are tried at least once
        action = e if e < len(Q) else np.argmax(Q + c * np.sqrt(np.log(e) / N))
        
        # Execute the action and observe the reward
        _, reward, _, _ = env.step(action)
        # Update the count and Q-value for the selected action
        N[action] += 1
        Q[action] = Q[action] + (reward - Q[action]) / N[action]
        
        # Store the current Q-values, reward, and action for this episode
        Qe[e] = Q
        returns[e] = reward
        actions[e] = action

    return name, returns, Qe, actions
```

```python
# This function implements the Thompson Sampling strategy for action selection in a given environment over a specified number of episodes.
# Thompson Sampling is a probabilistic method that balances exploration and exploitation by considering the uncertainty in the estimates of action values.
# It samples from a distribution over the estimated action values to decide which action to take, with the distribution becoming more peaked over the true action values as more is learned about the environment.

def thompson_sampling(env, 
                      alpha=1,
                      beta=0,
                      n_episodes=1000):
    # Initialize Q-values and counts for each action to zero
    Q = np.zeros((env.action_space.n), dtype=np.float64)
    N = np.zeros((env.action_space.n), dtype=np.int)
    
    # Arrays to store Q-values, returns, and actions for each episode
    Qe = np.empty((n_episodes, env.action_space.n), dtype=np.float64)
    returns = np.empty(n_episodes, dtype=np.float64)
    actions = np.empty(n_episodes, dtype=np.int)
    name = 'Thompson Sampling {}, {}'.format(alpha, beta)
    
    for e in tqdm(range(n_episodes), desc='Episodes for: ' + name, leave=False):
        # Sample from a normal distribution with mean Q and variance determined by alpha, N, and beta
        samples = np.random.normal(loc=Q, scale=alpha / (np.sqrt(N) + beta))
        action = np.argmax(samples)

        # Execute the action and observe the reward
        _, reward, _, _ = env.step(action)
        # Update the count and Q-value for the selected action
        N[action] += 1
        Q[action] = Q[action] + (reward - Q[action]) / N[action]

        # Store the current Q-values, reward, and action for this episode
        Qe[e] = Q
        returns[e] = reward
        actions[e] = action

    return name, returns, Qe, actions
```

```python
def b2_run_advanced_strategies_experiment(env_name='BanditTwoArmedUniform-v0'):
    results = {}
    experiments = [
        # baseline strategies 
        lambda env: pure_exploitation(env),
        lambda env: pure_exploration(env),

        # best from simple strategies
        lambda env: optimistic_initialization(env, 
                                              optimistic_estimate=1.0, 
                                              initial_count=10),
        lambda env: exp_dec_epsilon_greedy(env, 
                                           init_epsilon=0.3, 
                                           min_epsilon=0.0, 
                                           decay_ratio=0.3),

        # softmax
        lambda env: softmax(env, 
                            init_temp=float('inf'), 
                            min_temp=0.0, 
                            decay_ratio=0.005),
        lambda env: softmax(env, 
                            init_temp=100, 
                            min_temp=0.01, 
                            decay_ratio=0.01),

        # ucb
        lambda env: upper_confidence_bound(env, c=0.2),
        lambda env: upper_confidence_bound(env, c=0.5),

        # thompson sampling
        lambda env: thompson_sampling(env, alpha=1, beta=1),
        lambda env: thompson_sampling(env, alpha=0.5, beta=0.5),
    ]
    for env_seed in tqdm(SEEDS, desc='All experiments'):
        env = gym.make(env_name, seed=env_seed) ; env.reset()
        true_Q = np.array(env.env.p_dist * env.env.r_dist)
        opt_V = np.max(true_Q)
        for seed in tqdm(SEEDS, desc='All environments', leave=False):
            for experiment in tqdm(experiments, 
                                   desc='Experiments with seed {}'.format(seed), 
                                   leave=False):
                env.seed(seed) ; np.random.seed(seed) ; random.seed(seed)
                name, Re, Qe, Ae = experiment(env)
                Ae = np.expand_dims(Ae, -1)
                episode_mean_rew = np.cumsum(Re) / (np.arange(len(Re)) + 1)
                Q_selected = np.take_along_axis(
                    np.tile(true_Q, Ae.shape), Ae, axis=1).squeeze()
                regret = opt_V - Q_selected
                cum_regret = np.cumsum(regret)
                if name not in results.keys(): results[name] = {}
                if 'Re' not in results[name].keys(): results[name]['Re'] = []
                if 'Qe' not in results[name].keys(): results[name]['Qe'] = []
                if 'Ae' not in results[name].keys(): results[name]['Ae'] = []
                if 'cum_regret' not in results[name].keys(): 
                    results[name]['cum_regret'] = []
                if 'episode_mean_rew' not in results[name].keys(): 
                    results[name]['episode_mean_rew'] = []

                results[name]['Re'].append(Re)
                results[name]['Qe'].append(Qe)
                results[name]['Ae'].append(Ae)
                results[name]['cum_regret'].append(cum_regret)
                results[name]['episode_mean_rew'].append(episode_mean_rew)
    return results

b2_results_a = b2_run_advanced_strategies_experiment()
```
